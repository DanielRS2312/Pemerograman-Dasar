```python
# Nama: Daniel Reyvano Simanjuntak
# NIM: 4232511003
# Tanggal: 12 November 2026
# Mata Kuliah: RPE311 Pemrograman Dasar

# Soal 2
daya_kw = 450
jam = 8
energi_aktual = 7200
kapasitas_kw = 400

energi = daya_kw * jam
capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100

print(f"Energi harian = {energi} kWh")
print(f"Capacity factor = {capacity_factor:.2f}%")

# Cara kerja dan fungsi kode:
# Variabel daya_kw, jam, energi_aktual, dan kapasitas_kw berfungsi menyimpan nilai data operasional PLTD. Operator perkalian (*) digunakan untuk menghitung total energi harian dari daya dikali jam operasi.
# Selanjutnya, variabel capacity_factor menghitung persentase kinerja pembangkit dengan membagi energi aktual terhadap kapasitas maksimal dalam 24 jam lalu dikali 100. Terakhir, fungsi print() dan f-string menampilkan hasil ke layar, sedangkan format :.2f membatasi tampilan angka desimal menjadi dua angka di belakang koma.
```

```python
# Soal 6
suhu = float(input("Masukkan suhu uap (C): "))
tekanan = float(input("Masukkan tekanan (bar): "))

if suhu >= 550 or tekanan >= 14:
    print("STATUS: TRIP")
elif suhu > 520 and tekanan > 12:
    print("STATUS: WARNING")
else:
    print("STATUS: NORMAL")

# Cara kerja dan fungsi kode:
# Fungsi input() menerima masukan angka dari pengguna dan float() mengubahnya menjadi tipe data desimal agar bisa dibandingkan. Struktur percabangan if-elif-else dipakai untuk menentukan status boiler.
# Kondisi TRIP dicek paling pertama menggunakan operator logika 'or' agar jika salah satu batas bahaya tercapai, status langsung TRIP.
# Jika tidak, elif mengecek kondisi WARNING menggunakan operator 'and' (kedua syarat harus terpenuhi). Jika semua syarat di atas tidak terpenuhi, blok else akan mencetak NORMAL.
```

```python
# Soal 8
volume = 100

while volume > 0:
    volume = volume - 20
    print(f"Volume: {volume} liter")

    if volume == 20:
        print("WARNING: Hampir kosong!")

print("TANGKI KOSONG - Pengosongan berhenti.")

# Cara kerja dan fungsi kode:
# Variabel 'volume' menyimpan kapasitas awal tangki sebesar 100 liter.
# Perulangan while berfungsi mengulang proses pengurangan selama nilai volume masih lebih dari 0.
# Di setiap putaran, volume dikurangi 20 liter dan fungsi print() langsung menampilkan sisa volume terbaru.
# Di dalam perulangan terdapat percabangan if dengan operator == untuk mendeteksi saat volume tepat mencapai 20 liter dan mencetak peringatan.
# Setelah volume mencapai 0, perulangan otomatis berhenti dan baris print() terakhir dijalankan.
```

```python
# Soal 10
data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

titik = 1
valid = 0

for item in data:
    try:
        suhu = float(item)
        valid = valid + 1

        if suhu > 70.0:
            print(f"Warning: Titik {titik} suhu bearing tinggi!")

    except ValueError:
        print(f"Titik {titik}: data tidak valid - pemeriksaan dihentikan")
        break

    titik = titik + 1

print(f"Jumlah data valid: {valid}")

# Cara kerja dan fungsi kode:
# Perulangan for membaca setiap elemen teks di dalam list 'data' secara berurutan, sementara variabel 'titik' mencatat urutan sensor dan 'valid' menghitung data yang terbaca normal.
# Blok try berfungsi mencoba mengubah teks menjadi angka lewat float().
# Jika berhasil, hitungan 'valid' bertambah dan percabangan if mengecek apakah suhu melewati batas 70.0 untuk memunculkan peringatan.
# Jika ditemukan teks rusak seperti "N/A", blok except ValueError menangkap error konversi tersebut, mencetak pesan error, lalu perintah break langsung menghentikan perulangan saat itu juga.
```

```python
# Challenge - Pemantauan Beban Harian Generator
# Masalah: Memantau pembacaan beban daya (kW) tiap jam untuk mendeteksi beban
# berlebih (overload) sekaligus menangani data sensor yang error.

beban_harian = ["320.5", "410.0", "ERROR", "485.2", "390.0"]

jam = 1
total = 0.0
data_ok = 0

for b in beban_harian:
    try:
        daya = float(b)
        total = total + daya
        data_ok = data_ok + 1

        if daya >= 450.0:
            print(f"Jam {jam} ({daya} kW): OVERLOAD")
        elif daya >= 400.0:
            print(f"Jam {jam} ({daya} kW): Beban Tinggi")
        else:
            print(f"Jam {jam} ({daya} kW): Normal")

    except ValueError:
        print(f"Jam {jam}: Sensor error, data dilewati")

    jam = jam + 1

rata_rata = total / data_ok
print(f"Rata-rata beban terbaca: {rata_rata:.2f} kW")

# Cara kerja dan fungsi kode:
# List 'beban_harian' menyimpan riwayat pembacaan sensor daya generator setiap jam.
# Perulangan for memproses data satu per satu, dan blok try-except ValueError berfungsi mencegah program berhenti paksa saat membaca teks "ERROR" sehingga pemantauan tetap lanjut ke jam berikutnya.
# Untuk data yang valid, percabangan if-elif-else menentukan kategori beban (Normal, Beban Tinggi, atau Overload), sementara variabel 'total' dan 'data_ok' mengakumulasi nilai daya untuk menghitung rata-rata di akhir program.
```
