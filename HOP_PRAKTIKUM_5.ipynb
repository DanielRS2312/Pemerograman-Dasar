```python
# program untuk mengecek bonus dan diskon
total_belanja = input("Total belanja: Rp ")
# jumlah yang harus dibayar adalah berapa total belanjanya
# tapi kalau dapat diskon akan berkurang
bayar = int(total_belanja)
# jika dia belanja di atas 100rb maka berikan bonus dan diskon
if int(total_belanja) > 100000:
    print("Selamat Karena Belanja > Rp. 100.000 anda mendapat Voucher Makan ")
    print("dan diskon 5%")
    # hitung diskonnya
    diskon = int(total_belanja) * 5/100 #5%
    bayar = int(total_belanja) - diskon
# cetak struk
print("Total yang harus dibayar: Rp %s" % bayar)
print("Terima kasih sudah berbelanja di Toko Kami")
print("Kami Tunggu kunjungan anda selanjutnya.")
```

```python
# Contoh Pengecekan Syarat Pembuatan SIM Berdasarkan Umur

umur = input("Berapa umur kamu: ")
if int(umur) >= 18:
    print("Kamu boleh membuat SIM")
else:
    print("Kamu belum boleh membuat SIM")
```

```python
# Contoh Pengecekan Bilangan Genap atau Ganjil Menggunakan Modulus

angka = int(input("Masukkan bilangan: "))

if angka % 2 == 0:
    print("Bilangan Genap")
else:
    print("Bilangan Ganjil")
```

```python
# Contoh Program Diskon Belanja Berdasarkan Batas Nominal

total_belanja = 250000 # asumsikan total belanja di atas 20rb (atau sesuai batas ketentuan)

if total_belanja > 200000:
    diskon = total_belanja * 0.10
    total_bayar = total_belanja - diskon
    print(f"Anda mendapat diskon! Total bayar: Rp {total_bayar}")
else:
    print(f"Total bayar: Rp {total_belanja}")
```

```python
# Contoh Penilaian Nilai Angka Berdasarkan Rentang Huruf Mutu

nilai = 80
if nilai >= 85:
    print("Luar biasa! Nilai A")
elif nilai >= 70:
    print("Bagus! Nilai B")
elif nilai >= 60:
    print("Cukup. Nilai C")
else:
    print("Nilai kurang, perlu belajar lebih giat.")
```

```python
# Menentukan jenis diskon berdasarkan hari
hari: str = "Sabtu"

if hari == "Sabtu" or hari == "Sunday": # (atau hari == "Minggu")
    diskon = float(0.20) # Diskon 20%
    print(f"Selamat! Anda mendapat diskon akhir pekan sebesar {diskon*100}%.")
elif hari == "Jumat":
    diskon = float(0.15) # Diskon 15%
    print(f"Diskon spesial Jumat! Anda mendapat potongan {diskon*100}%.")
else:
    diskon = float(0.05) # Diskon 5%
    print(f"Anda mendapat diskon hari biasa sebesar {diskon*100}%.")
```

```python
# Contoh Program Menu Konversi Mata Uang Interaktif

rupiah = int(input("Masukkan jumlah uang (Rp): "))
print("1. Dollar (USD)")
print("2. Euro (EUR)")
print("3. Yen (JPY)")
pilih = int(input("Pilih konversi (1/2/3): "))

if pilih == 1:
    print("Dalam Dollar:", rupiah / 15000)
elif pilih == 2:
    print("Dalam Euro:", rupiah / 16000)
elif pilih == 3:
    print("Dalam Yen:", rupiah / 120)
else:
    print("Pilihan tidak valid")
```
