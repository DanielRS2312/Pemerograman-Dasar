```python
# B.1 Soal Umum
# 1. Program operasi aritmatika lengkap (+, -, *, /, //, %)

a = int(input("Masukkan bilangan bulat pertama: "))
b = int(input("Masukkan bilangan bulat kedua: "))

print("Penjumlahan:", a + b)
print("Pengurangan:", a - b)
print("Perkalian:", a * b)
print("Pembagian:", a / b)
print("Pembagian bulat:", a // b)
print("Modulus:", a % b)
```

```python
# B.1 Soal Umum
# 3. Program cek boleh mengemudi (menggunakan operator logika dan tanda kurung)

umur = int(input("Masukkan umur Anda: "))
punya_sim = input("Apakah punya SIM? (True/False): ") == "True"

boleh_mengemudi = (umur >= 17) and punya_sim
print(f"Boleh mengemudi sendiri? {boleh_mengemudi}")
```

```python
# B.2 Soal Konteks Energi
# 4. Program menghitung efisiensi termal unit pembangkit

energi_input = float(input("Masukkan nilai energi input (dalam MJ): "))
energi_output = float(input("Masukkan nilai energi output (dalam MJ): "))

# Menggunakan tanda kurung dan f-string dengan pembulatan 2 angka di belakang koma
efisiensi = (energi_output / energi_input) * 100
print(f"Efisiensi termal unit pembangkit adalah {efisiensi:.2f}%")
```

```python
# B.2 Soal Konteks Energi
# 5. Program cek kondisi "perlu maintenance" berdasarkan suhu boiler dan beban operasi

suhu_boiler = float(input("Masukkan suhu boiler (°C): "))
beban_operasi = float(input("Masukkan beban operasi (%): "))

# True jika suhu boiler > 550 atau beban operasi > 95
perlu_maintenance = (suhu_boiler > 550) or (beban_operasi > 95)

print(f"Status unit pembangkit perlu maintenance: {perlu_maintenance}")
```

```python
# Soal 1: Belanja Buku
jumlah_buku = 7
harga_satuan = 18500
total_belanja = jumlah_buku * harga_satuan

diskon = 0.1 * total_belanja
jumlah_bayar = total_belanja - diskon

uang_budi = 150000
kembalian = uang_budi - jumlah_bayar

# Patungan 4 orang secara adil (bilangan bulat dan sisa)
per_orang = int(jumlah_bayar // 4)
sisa_rupiah = int(jumlah_bayar % 4)

uang_cukup = uang_budi >= jumlah_bayar

print("Total    :", total_belanja)
print("Diskon   :", diskon)
print("Bayar    :", jumlah_bayar)
print("Kembalian:", kembalian)
print("Per orang:", per_orang)
print("Sisa     :", sisa_rupiah)
print("Uang cukup?:", uang_cukup)
```

```python
# Soal 2: PLTS Atap Rumah
jumlah_panel = 12
daya_panel_watt = 450
penyinaran = 5
efisiensi = 0.8
kebutuhan_listrik = 25
tarif_kwh = 1444.70

# 1. Energi harian (kWh) -> Watt ke kWh dibagi 1000
energi_harian = (jumlah_panel * daya_panel_watt * penyinaran * efisiensi) / 1000

# 2. Energi bulanan (30 hari) dan nilai rupiah
energi_bulanan = energi_harian * 30
nilai_rupiah = energi_bulanan * tarif_kwh

# 3. Kekurangan energi per hari
kekurangan = kebutuhan_listrik - energi_harian

# 4. Apakah sudah mencukupi?
mencukupi = energi_harian >= kebutuhan_listrik

print(f"Energi harian (kWh): {round(energi_harian, 1)}")
print(f"Energi bulanan (kwh): {round(energi_bulanan, 1)}")
print(f"Nilai listrik (Rp)  : {round(nilai_rupiah, 1)}")
print(f"Kekurangan (kWh)    : {round(kekurangan, 1)}")
print(f"Mencukupi?          : {mencukupi}")
```

```python
# Soal 3: Generator Diesel dan Rugi Kabel
kapasitas_generator = 200  # kW
jam_operasi = 8
konsumsi_solar_per_kwh = 0.25
harga_solar = 13500
kapasitas_tangki = 1000

arus = 20
hambatan = 0.5

# 1. Energi yang dibangkitkan per hari (kWh) = kW * jam
energi_dibangkitkan = kapasitas_generator * jam_operasi

# 2. Kebutuhan solar per hari dan biayanya
solar_per_hari = energi_dibangkitkan * konsumsi_solar_per_kwh
biaya_solar = solar_per_hari * harga_solar

# 3. Hari penuh tangki cukup dan sisa solar
hari_cukup = kapasitas_tangki // solar_per_hari
sisa_solar = kapasitas_tangki % solar_per_hari

# 4. Rugi daya pada kabel (P = I^2 * R)
rugi_kabel = (arus ** 2) * hambatan

# 5. Apakah rugi kabel tinggi? (> 150 W dan arus minimal 20 A)
rugi_tinggi = (rugi_kabel > 150) and (arus >= 20)

print("Energi (kwh)  :", energi_dibangkitkan)
print("Solar (liter) :", solar_per_hari)
print("Biaya (Rp)    :", biaya_solar)
print("Tangki cukup  :", hari_cukup, "hari")
print("Sisa solar (L):", sisa_solar)
print("Rugi kabel (W):", rugi_kabel)
print("Rugi tinggi?  :", rugi_tinggi)
```
