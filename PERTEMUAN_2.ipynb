```python
# === LATIHAN 1 ===
# Panel surya: tegangan 18 V, arus 6 A, jenis sel Monocrystalline
# Sesuai aturan penulisan PEP 8

tegangan_panel = 18
arus_panel = 6
jenis_sel = "Monocrystalline"

# Cetak semua nilai
print("=== Spesifikasi Panel Surya ===")
print(f"Tegangan : {tegangan_panel} volt")
print(f"Arus     : {arus_panel} ampere")
print(f"Jenis sel: {jenis_sel}")
```

```python
# === LATIHAN 2 ===
jumlah_komponen = 10
kapasitansi_uf = 4.7
status_aktif = True
nama_komponen = "Resistor"

print("=== Spesifikasi Komponen ===")
print(f"Jumlah komponen : {jumlah_komponen} | Tipe: {type(jumlah_komponen)}")
print(f"Kapasitansi     : {kapasitansi_uf} uF | Tipe: {type(kapasitansi_uf)}")
print(f"Status aktif    : {status_aktif} | Tipe: {type(status_aktif)}")
print(f"Nama komponen   : {nama_komponen} | Tipe: {type(nama_komponen)}")
```

```python
# === LATIHAN 3 ===
nama_unit = "PLTS Nunukan"
jumlah_panel = 240
tegangan_keluaran = 48.5
sedang_beroperasi = True

print("=== Data Unit PLTS ===")
print(f"Nama unit         : {nama_unit} | Tipe: {type(nama_unit)}")
print(f"Jumlah panel      : {jumlah_panel} | Tipe: {type(jumlah_panel)}")
print(f"Tegangan keluaran : {tegangan_keluaran} volt | Tipe: {type(tegangan_keluaran)}")
print(f"Sedang beroperasi : {sedang_beroperasi} | Tipe: {type(sedang_beroperasi)}")
```

```python
# === LATIHAN 4 ===
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

print("=== Data Tegangan Harian ===")
# (a) Cetak pembacaan pertama dan terakhir memakai indeks
print(f"Pembacaan pertama  : {tegangan_harian[0]} volt")
print(f"Pembacaan terakhir : {tegangan_harian[-1]} volt")

# (b) Tambahkan satu pembacaan baru 222.0 dengan append()
tegangan_harian.append(222.0)

# (c) Cetak jumlah total data dengan len()
print(f"Jumlah total data  : {len(tegangan_harian)} data")
```

```python
# === LATIHAN 5 ===
motor_listrik = {
    "nama": "Motor Induksi AC",
    "daya_kW": 5.5,
    "tegangan_V": 380,
    "rpm": 1450
}

print("=== Spesifikasi Motor Listrik ===")

# Cetak salah satu nilai lewat key-nya
print(f"Daya motor : {motor_listrik['daya_kW']} kW")

# Ubah nilai rpm (pakai =)
motor_listrik["rpm"] = 1500

# Cetak ulang seluruh dictionary
print(f"Data motor : {motor_listrik}")
```

```python
# === LATIHAN 6 ===
pembangkit = [
    {"nama": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_MW": 80, "status": "pemeliharaan"}
]

print("=== Data Pembangkit Listrik ===")

# 1. Cetak nama unit kedua
print(f"Nama unit kedua : {pembangkit[1]['nama']}")

# 2. Ubah status salah satu unit (misal Unit-3 jadi 'operasi')
pembangkit[2]["status"] = "operasi"

# 3. Tambahkan satu unit baru ke list dengan append()
pembangkit.append({"nama": "Unit-4", "kapasitas_MW": 120, "status": "pemeliharaan"})

# 4. Cetak seluruh data
print(f"Seluruh data    : {pembangkit}")
```
